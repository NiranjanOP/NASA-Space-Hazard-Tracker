```sql
from pyspark.sql.functions import col

df_raw = spark.table("nasa_analytics.bronze.donki_raw")
```

```sql
df_structured = df_raw.select(
    col("messageID").alias("message_id"),
    col("messageType").alias("message_type"),
    col("messageIssueTime").alias("issue_time"),
    col("messageBody").alias("message_body"),
    col("messageURL").alias("message_url")
)

# Save to silver
df_structured.write.mode("overwrite").saveAsTable("nasa_analytics.silver.donki_notifications")
```

```sql
# Solar flare alerts
df_structured.filter(col("message_type") == "FLR") \
    .write.mode("overwrite").saveAsTable("nasa_analytics.silver.donki_flare_alerts")

# Coronal mass ejections
df_structured.filter(col("message_type") == "CME") \
    .write.mode("overwrite").saveAsTable("nasa_analytics.silver.donki_cme_alerts")

# Geomagnetic storms
df_structured.filter(col("message_type") == "GST") \
    .write.mode("overwrite").saveAsTable("nasa_analytics.silver.donki_geomagnetic_alerts")
```
