```sql
from pyspark.sql.functions import *
import requests
import json

# API Key
api_key = "YOUR_API_KEY"  # replace with your actual key

# API Endpoint (example: feed for a given date range)
url = f"https://api.nasa.gov/neo/rest/v1/feed?start_date=2025-09-20&end_date=2025-09-22&api_key={api_key}"

# Fetch data
response = requests.get(url)
data = response.json()
```

```sql
# Convert to Spark DF
df = spark.createDataFrame([json.loads(json.dumps(data))])

# Write to bronze schema as raw JSON
df.write.mode("overwrite").saveAsTable("nasa_analytics.bronze.neows_raw")
```

```sql
USE CATALOG nasa_analytics;
SELECT * FROM bronze.neows_raw LIMIT 10;
```
