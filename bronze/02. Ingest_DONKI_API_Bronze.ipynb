```sql
import requests
import json
from pyspark.sql.functions import *

# API Key
api_key = "YOUR_API_KEY"

# Example DONKI endpoint → notifications
url = f"https://api.nasa.gov/DONKI/notifications?startDate=2025-09-01&endDate=2025-09-25&type=all&api_key={api_key}"

# Fetch
response = requests.get(url)
data = response.json()
```

```sql
--Convert list of JSON objects → Spark DataFrame
df = spark.createDataFrame(data)

# Write as raw JSON to bronze
df.write.mode("overwrite").saveAsTable("nasa_analytics.bronze.donki_raw")
```

```sql
USE CATALOG nasa_analytics;
SELECT * FROM bronze.donki_raw LIMIT 20;
```
