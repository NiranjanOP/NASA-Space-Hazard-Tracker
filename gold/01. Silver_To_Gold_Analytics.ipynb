```sql
USE CATALOG nasa_analytics;
```

```sql
CREATE OR REPLACE TABLE gold.hazardous_asteroids_daily AS
SELECT
    approach_date,
    COUNT(*) AS total_asteroids,
    SUM(CASE WHEN is_hazardous THEN 1 ELSE 0 END) AS hazardous_count,
    MIN(miss_distance_km) AS closest_miss_distance_km,
    MAX(velocity_kph) AS max_velocity_kph
FROM silver.neows_asteroids
GROUP BY approach_date
ORDER BY approach_date;
```

```sql
CREATE OR REPLACE TABLE gold.largest_hazardous_asteroids AS
SELECT
    approach_date,
    asteroid_id,
    name,
    diameter_max_km,
    is_hazardous,
    miss_distance_km
FROM silver.neows_asteroids
WHERE is_hazardous = TRUE
ORDER BY diameter_max_km DESC
LIMIT 50;
```

```sql
CREATE OR REPLACE VIEW gold.space_hazard_overview AS
SELECT
    ha.approach_date,
    ha.total_asteroids,
    ha.hazardous_count,
    ha.closest_miss_distance_km,
    ha.max_velocity_kph,
    lha.name AS largest_hazard_name,
    lha.diameter_max_km AS largest_hazard_diameter
FROM gold.hazardous_asteroids_daily ha
LEFT JOIN gold.largest_hazardous_asteroids lha
  ON ha.approach_date = lha.approach_date;
```

```sql
SELECT * FROM gold.space_hazard_overview
LIMIT 20;
```
